**1.** Monte um passo a passo para o algoritmo RF

- Semelhante ao *Bagging*, o *Random Forest* consiste nos seguintes passos:
1. *Bootstrap + Feature Selection:* : Assim como no Bagging, o Random Forest cria amostras aleatórias com reposição a partir do conjunto de dados original. Porém, em cada amostra, apenas um conjunto limitado e aleatório de variáveis é escolhido (feature selection). Em problemas de classificação, é comum escolher a raiz quadrada do número total de variáveis, enquanto em problemas de regressão, normalmente se utiliza um terço das variáveis.
2. *Modelagem com árvores de decisão:* Nesta fase, um modelo de Machine Learning, especificamente uma árvore de decisão, é treinado separadamente em cada amostra bootstrap utilizando as variáveis aleatórias definidas na etapa anterior.
3. *Agregação:*  Por fim, os resultados de cada árvore de decisão são combinados para gerar a previsão final. Em problemas de classificação, a combinação geralmente é feita pelo voto da maioria, onde a classe mais frequentemente prevista pelas árvores é escolhida como a resposta final. Em problemas de regressão, a média das previsões das árvores é calculada para determinar o resultado final.

**2.** Explique com suas palavras o Random Forest

O Random Forest pode ser visto como uma evolução do Bootstrap Aggregating (ou Bagging), oferecendo melhorias em desempenho e precisão. Similar ao Bagging, o Random Forest é uma técnica de combinação de modelos de Machine Learning, onde cada modelo é treinado em versões modificadas do conjunto de dados original. Essas versões, conhecidas como amostras bootstrap, são subconjuntos aleatórios do conjunto de dados original, com possíveis repetições, mantendo o mesmo número de observações. Entretanto, o Random Forest se distingue ao selecionar aleatoriamente um número limitado de variáveis em cada modelo, especialmente em árvores de decisão. Esse método visa diminuir a variância dos resultados e reduzir o risco de overfitting. A previsão final é obtida combinando os modelos de forma semelhante ao Bagging, utilizando a média para regressão e a votação para classificação.

**3.** Qual a diferença entre Bagging e Random Forest?

A diferença entre Bagging e Random Forest está no desempenho e na capacidade de reduzir a variância dos resultados dos modelos. Bagging é uma técnica mais ampla de combinação de modelos de Machine Learning que utiliza amostragem aleatória, enquanto o Random Forest é uma aplicação mais específica desse método, que faz uso de árvores de decisão. No Random Forest, cada modelo é treinado em subconjuntos aleatórios do conjunto de dados original, com uma seleção aleatória e reduzida de variáveis em cada modelo. Essa estratégia aumenta a robustez do modelo e, ao combinar os resultados, a previsão final tende a ser mais precisa do que no Bagging. Assim, o Random Forest é uma versão aprimorada do Bagging, oferecendo um desempenho superior e uma redução adicional na variância dos resultados. O Random Forest se destaca em relação ao Bagging, pois as árvores amostradas apresentam menor correlação, tornando-as mais independentes.
